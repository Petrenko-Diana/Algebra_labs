Розглянемо групу S4 та її підгрупи:

(a) H1 = &lt;(1,2,3,4)&gt;

(b) H2 = &lt;(1,3), (1,2,3,4)&gt;

(c) H3 = K4

Для кожної підгрупи H виконайте наступні кроки:

Обчисліть порядок підгрупи H та її індекс у групі [S4 : H].

Знайдіть ліві та праві класи суміжності підгрупи H у S4.

Визначте, чи є H нормальною підгрупою S4.

Якщо H є нормальною, то побудуйте факторгрупу G/H, її таблицю Келі та визначіть структуру.

Якщо H не є нормальною, то знайдіть такий елемент a ∈ S4, що aH ≠ Ha.


```sage
G = SymmetricGroup(4)

H1 = G.subgroup([G("(1,2,3,4)")])
H2 = G.subgroup([G("(1,3)"), G("(1,2,3,4)")])
H3 = KleinFourGroup()  # Або G.subgroup([G("(1,2)(3,4)"), G("(1,3)(2,4)")])

subgroups = [("(a) H1 = <(1,2,3,4)>", H1), 
             ("(b) H2 = <(1,3), (1,2,3,4)>", H2), 
             ("(c) H3 = K4", H3)]

for name, H in subgroups:
    print("=" * 60)
    print(f"Підгрупа {name}")
    
    # 1. Порядок та індекс
    order_H = H.order()
    index_H = G.order() // order_H
    print(f"1. Порядок |H| = {order_H}, індекс [S4 : H] = {index_H}")
    
    # 2. Ліві та праві класи суміжності
    left_cosets = G.cosets(H, side='left')
    right_cosets = G.cosets(H, side='right')
    print(f"2. Ліві класи суміжності ({len(left_cosets)} шт.):")
    for c in left_cosets:
        print("   ", c)
    print(f"   Праві класи суміжності ({len(right_cosets)} шт.):")
    for c in right_cosets:
        print("   ", c)
        
    # 3. Перевірка на нормальність
    is_norm = H.is_normal(G)
    print(f"3. Чи є H нормальною підгрупою в S4? -> {is_norm}")
    
    # 4 або 5. Факторгрупа або контрприклад
    if is_norm:
        Q = G.quotient(H)
        print(f"4. Факторгрупа S4/H має порядок {Q.order()} і ізоморфна: {Q.structure_description()}")
        print("   Таблиця Келі факторгрупи:")
        print(Q.cayley_table())
    else:
        for a in G:
            left_c = Set([a * h for h in H])
            right_c = Set([h * a for h in H])
            if left_c != right_c:
                print(f"5. Підгрупа не нормальна. Контрприклад a = {a}:")
                print(f"   aH = {list(left_c)}")
                print(f"   Ha = {list(right_c)}")
                break
```

Розглянемо групу GL2(Z3) та множину прямих M = {l1, l2, l3, l4} у просторі Z3^2, що визначаються парами пропорційних ненульових векторів:

l1 = {(1, 0), (2, 0)}

l2 = {(0, 1), (0, 2)}

l3 = {(1, 1), (2, 2)}

l4 = {(1, 2), (2, 1)}

Побудуйте групу GL2(Z3).

Задайте відображення phi: GL2(Z3) -&gt; S4, яке переводить матрицю у перестановку, що описує її дію на прямі з M.

Перевірте, що phi є гомоморфізмом груп.

Обчисліть ядро N = ker(phi) та образ Im(phi).

Побудуйте факторгрупу GL2(Z3) / N та порівняйте її з образом Im(phi).


```sage
F, G_mat = GF(3), GL(2, GF(3))
lines = [{tuple(vector(F, [1, 0])), tuple(vector(F, [2, 0]))},
         {tuple(vector(F, [0, 1])), tuple(vector(F, [0, 2]))},
         {tuple(vector(F, [1, 1])), tuple(vector(F, [2, 2]))},
         {tuple(vector(F, [1, 2])), tuple(vector(F, [2, 1]))}]
S4 = SymmetricGroup(4)

def act(A):
    imgs = []
    for base in lines:
        tr = {tuple(A * vector(F, v)) for v in base}
        for idx, l in enumerate(lines):
            if tr == l:
                imgs.append(idx + 1)
                break
    return S4(imgs)

ker = [A for A in G_mat if act(A) == S4.one()]
im = S4.subgroup([act(A) for A in G_mat])
print(f"GL2(Z3) -> Ядро: {len(ker)}, Образ: {im.order()}, Структура: {im.structure_description()}")
```

```sage
# 1. Задаємо першу твірну a у вигляді циклів

a_cycles = [
    (1,60), (2,72), (3,81), (4,43), (5,11),
    (6,87), (7,34), (9,63), (12,46), (13,28),
    (14,71), (15,42), (16,97), (18,57), (19,52),
    (21,32), (23,47), (24,54), (25,83), (26,78),
    (29,89), (30,39), (33,61), (35,56), (37,67),
    (44,76), (45,88), (48,59), (49,86), (50,74),
    (51,66), (53,99), (55,75), (62,73), (65,79),
    (68,82), (77,92), (84,90), (85,98), (94,100)
]

# 2. Задаємо другу твірну b у вигляді циклів

b_cycles = [
    (1,86,13,10,47),
    (2,53,30,8,38),
    (3,40,48,25,17),
    (4,29,92,88,43),
    (5,98,66,54,65),
    (6,27,51,73,24),
    (7,83,16,20,28),
    (9,23,89,95,61),
    (11,42,46,91,32),
    (12,14,81,55,68),
    (15,90,31,56,37),
    (18,69,45,84,76),
    (19,59,79,35,93),
    (21,22,64,39,100),
    (26,58,96,85,77),
    (33,52,94,75,44),
    (34,62,87,78,50),
    (36,82,60,74,72),
    (41,80,70,49,67),
    (57,63,71,99,97)
]

# 3. Створюємо симетричну групу S100
# та перетворюємо цикли на перестановки

S = SymmetricGroup(100)

a = S(a_cycles)
b = S(b_cycles)

print("Твірна a:", a)
print("Твірна b:", b)

# 4. Перевіряємо порядки твірних

print("\nПорядок a:", a.order())
print("Порядок b:", b.order())
print("Порядок ab:", (a*b).order())
print("Порядок ab^2:", (a*b^2).order())

# Додаткова перевірка правильності твірних

assert a.order() == 2
assert b.order() == 5
assert (a*b).order() == 11
assert (a*b^2).order() == 10

print("\nУсі перевірки твірних пройдено!")

# 5. Будуємо підгрупу H, породжену a та b

H = PermutationGroup([a, b])

print("\nГрупа H = <a,b>")
print("Степінь групи:", H.degree())

# 6. Обчислюємо порядок групи

print("\nПорядок групи H:", H.order())

# 7. Знаходимо всі нормальні підгрупи

normal_subgroups = H.normal_subgroups()

print("\nКількість нормальних підгруп:", len(normal_subgroups))

print("\nНормальні підгрупи:")

for i, K in enumerate(normal_subgroups):
    print("Підгрупа", i+1)
    print("Порядок:", K.order())

# 8. Перевіряємо, чи є група простою

print("\nЧи є група простою?", H.is_simple())

# 9. Знаходимо центр групи

Z = H.center()

print("\nЦентр групи:", Z)
print("Порядок центра:", Z.order())

# 10. Підсумкова перевірка

print("\n----------- ВИСНОВОК ----------")

if H.is_simple():
    print("Група H є простою.")
else:
    print("Група H не є простою.")

print("Порядок групи:", H.order())
print("Кількість нормальних підгруп:", len(normal_subgroups))
```
