```sage
# Z*14
R = Integers(14)
units = [R(x) for x in range(14) if gcd(x, 14) == 1]

# У групі одиниць за модулем 14 елемент 3 породжує всі елементи.
generator = R(3)
powers = []
x = R(1)
while x not in powers:
    powers.append(x)
    x *= generator
is_cyclic = set(map(int, powers)) == set(map(int, units))

print("--- Група Z_14* (мультиплікативна) ---")

# 2. Порядок групи
print("2. Порядок групи:", len(units))

# 3. Система твірних
print("3. Система твірних:", [int(generator)] if is_cyclic else "нециклічна група")

# 4. Елементи групи
print("4. Елементи групи (оригінальні числа):", [int(x) for x in units])

# 5. Таблиця Келі для множення за модулем 14
print("5. Таблиця Келі:")
print("   * | " + " ".join(f"{int(b):>2}" for b in units))
print("-----+" + "---" * len(units))
for a in units:
    print(f"{int(a):>4} | " + " ".join(f"{int(a * b):>2}" for b in units))

# 6. Чи є абелевою
print("6. Чи є абелевою:", all(a * b == b * a for a in units for b in units))

# 7. Структура групи
print("7. Структура групи:", f"C{len(units)}" if is_cyclic else "нециклічна абелева група")
```

2\. Працюємо з елементами груп


```sage
G = Integers(10)
print("--- Робота з елементами Z_10 ---")
# Операція в Z_10 — додавання
a, b = G(3), G(4)
print(f"2. Елементи: a = {a}, b = {b}. Операція (a + b) = {a + b}")

# Обернені елементи для додавання — протилежні елементи
print(f"3. Обернені: -a = {-a}, -b = {-b}. Перевірка a+(-a) = {a + (-a)}")

# Підгрупа адитивної групи, породжена a та b:
# усі комбінації i*a + j*b достатньо перевірити для i,j від 0 до 9.
H = {G(i * int(a) + j * int(b)) for i in range(10) for j in range(10)}
print(f"4. Порядок підгрупи <a, b>: {len(H)}")
print("   Елементи підгрупи:", sorted(H))

# Елементи, які є оберненими до самих себе в адитивній групі
self_inverse = [x for x in G if (x + x) == 0]
print("5. Обернені самі до себе:", self_inverse)

# Перевірка на абелевість
print("6. Група є абелевою:", all(x + y == y + x for x in G for y in G))
```

```sage
G = Integers(10).unit_group()
print("--- Робота з елементами Z_10* ---")
# 2. Вибір елементів
a, b = G.gens()[0], list(G)[-1]
print(f"2. Елементи: a = {a}, b = {b}. Добуток a * b = {a * b}")
# 3. Обернені
print(f"3. Обернені: a^-1 = {a^-1}, b^-1 = {b^-1}. Перевірка a*a^-1 = {a * a^-1}")
# 4. Підгрупа
H = G.subgroup([a, b])
print(f"4. Порядок підгрупи <a, b>: {H.order()}")
# 5. Обернені самі до себе
self_inverse = [x for x in G if x^2 == G.one()]
print("5. Обернені самі до себе:", self_inverse)
print("6. Група є абелевою.")
```

```sage
G = AlternatingGroup(4)
print("--- Робота з елементами A_4 ---")
a, b = G.gens()[0], G.gens()[1]
print(f"2. Елементи: a = {a}, b = {b}. Добуток a * b = {a * b}")
print(f"3. Обернені: a^-1 = {a^-1}, b^-1 = {b^-1}. Перевірка a*a^-1 = {a * a^-1}")
H = G.subgroup([a, b])
print(f"4. Порядок підгрупи <a, b>: {H.order()}")
self_inverse = [x for x in G if x^2 == G.one()]
print("5. Обернені самі до себе:", self_inverse)
# 6. Неабелева група - пошук елементів, що не комутують
for x in G:
    for y in G:
        if x*y != y*x:
            print(f"6. Елементи не комутують: x = {x}, y = {y}. Бо x*y = {x*y}, а y*x = {y*x}")
            break
    if x*y != y*x: break
```

```sage
G = GL(2, GF(3))
print("--- Робота з елементами GL_2(Z_3) ---")
a, b = G.gens()[0], G.gens()[1]
print("2. Елемент a:\n", a, "\nЕлемент b:\n", b, "\nДобуток a * b:\n", a * b)
print("3. Перевірка оберненого a * a^-1:\n", a * a^-1)
H = G.subgroup([a, b])
print(f"4. Порядок підгрупи <a, b>: {H.order()}")
self_inverse = [x for x in G if x^2 == G.one()]
print(f"5. Кількість елементів обернених самі до себе: {len(self_inverse)}")
# 6. Неабелеві елементи
for x in G:
    for y in G:
        if x*y != y*x:
            print("6. Знайдено елементи, що не комутують:\nx =\n", x, "\ny =\n", y)
            break
    if x*y != y*x: break
```

3\. Підгрупи та їхні властивості


```sage
G = CyclicPermutationGroup(24)
print("--- Підгрупи Z_24 ---")
subgroups = G.subgroups()
print("2. Список усіх підгруп (всього):", len(subgroups))
print("3. Кількість підгруп з точністю до ізоморфізму:", len(G.conjugacy_classes_subgroups()))
orders = [H.order() for H in subgroups]
print("4. Порядки всіх підгруп:", sorted(orders))
check = all(G.order() % o == 0 for o in orders)
print("5. Чи порядок кожної підгрупи ділить порядок групи?", check)
```

4\. Графи Келі


```sage
G = DihedralGroup(4)
print("--- Граф Келі D_4 ---")
# 1. Побудова графа
cayley = G.cayley_graph()
# 2. Степінь вершин
print("2. Степінь вершин графа Келі:", cayley.degree(cayley.vertices()[0]))
# 3. Вершинно-транзитивний
print("3. Чи є вершинно-транзитивним:", cayley.is_vertex_transitive())
# 4. Гамільтонів цикл
print("4. Чи містить гамільтонів цикл:", cayley.is_hamiltonian())
cayley.show() # Візуалізація
```
