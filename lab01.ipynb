```sage
# а) оголосіть змінні x=10 і y=15 та обчисліть їхню суму і добуток
x_val = 10
y_val = 15
print("а) Сума:", x_val + y_val, "| Добуток:", x_val * y_val)

# Очистимо x та оголосимо z для символьних обчислень
var('x z')

# б) обчисліть вираз 3z^2 + 2z - 5
expr_z = 3*z^2 + 2*z - 5
print("б) Вираз із z:", expr_z)

# в) спростіть вираз 2x^2 + 4x - 2x^2 + x
expr_sim = 2*x^2 + 4*x - 2*x^2 + x
print("в) Спрощений вираз:", expr_sim.simplify_full())

# г) розкрийте дужки у виразі (x + 2)(x - 3)
expr_expand = (x + 2) * (x - 3)
print("г) Розкриті дужки:", expr_expand.expand())

# д) розкладіть на множники тричлен x^2 - 9x + 20
expr_factor = x^2 - 9*x + 20
print("д) Розкладання на множники:", expr_factor.factor())

# е) розв'яжіть рівняння x^2 - 4x + 3 = 0
eq = x^2 - 4*x + 3 == 0
print("е) Корені рівняння:", solve(eq, x))

```

```sage
var('x')

# а) графік функції f(x) = x^2 - 4x + 3 на [-5, 5]
p1 = plot(x^2 - 4*x + 3, (x, -5, 5), title="Графік а)")
p1.show()

# б) похідна функції f(x) = x^3 + 5x^2 - 2x + 7
f_der = x^3 + 5*x^2 - 2*x + 7
print("б) Похідна:", diff(f_der, x))

# в) невизначений інтеграл функції 2x^3 - x^2 + 5
f_int = 2*x^3 - x^2 + 5
print("в) Невизначений інтеграл:", integral(f_int, x))

# г) визначений інтеграл від 1 до 4 для x^2
print("г) Визначений інтеграл:", integral(x^2, x, 1, 4))

# д) функція f(x) = 3x^2 - 2x + 1, обчислити f(5)
f(x) = 3*x^2 - 2*x + 1
print("д) f(5) =", f(5))

# е) два графіки на одній площині на [-3,3]
p2 = plot(x^2, (x, -3, 3), color="blue", legend_label="f(x)=x^2")
p3 = plot(-x^2 + 5, (x, -3, 3), color="red", legend_label="g(x)=-x^2+5")
(p2 + p3).show()
```
